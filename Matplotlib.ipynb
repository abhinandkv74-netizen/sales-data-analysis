## Histogram
 used to show the distributoion of numerical data by grouping values in to intervels called bins

```python
marks=[30,35,40,45,50,55,60,65,70]
```

```python
plt.hist(marks,bins=5,edgecolor="black")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.title(" mark detiles")
plt.show()
```
